```python
from pathlib import Path

import torch
from IPython.display import display

from detect import Detect
from main import load_net

images = Path("presentation_images")
scales = tuple(0.5 * 0.8**i for i in range(8))

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
net = load_net("FCNet", "FCNet.pt").to(device)

detector = Detect(
    net=net,
    scaling_sizes=scales,
    certainty_filter=1.5,
    knn=0,
    thickness=6,
    debug=False,
    step_size=2**5
)
```

```python
extensions = {".jpg", ".jpeg", ".png", ".webp", ".gif"}

for path in sorted(images.iterdir()):
    if path.is_file() and path.suffix.lower() in extensions:
        print(path.name)
        result = detector.detect_faces(path)
        display(result)
```
